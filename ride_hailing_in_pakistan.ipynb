```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\rahma khawaja\Downloads\ride_hailing_dataset.csv")
```

```python
df.head()
```

```python
avg_fare_by_service = df.groupby('service_type')['fare'].mean()
print("Q1 - Average fare by service type:")
print(round(avg_fare_by_service, 2))
```

```python
top_city = df['city'].value_counts().idxmax()
print("\nQ2 - City with the most trips:", top_city)
```

```python
avg_distance = np.nanmean(df['distance_km'])
print("\nQ3 - Average distance per trip:", round(avg_distance, 2))
```

```python
ratings_by_gender = df.groupby('Gender')['Rating'].mean()
print("\nQ4 - Average trip rating by gender:")
print(round(ratings_by_gender, 2))
```
